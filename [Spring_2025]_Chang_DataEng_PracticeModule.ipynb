## **Assignment**<font color="#F1A006">  ▝</font><font color="#707DA0">▖</font>**1**<font color="#F1A006">▝</font><font color="#707DA0">▖   </font>

The TMDB Movie Database is an up-to-date comprehensive dataset containing information about 700,000 movies.
[Source: Accessed May 5, 2024](https://www.kaggle.com/datasets/akshaypawar7/millions-of-movies)

For the assignments, the data includes only the following columns:

- `title`: Title of the movie
-  `release_date`: Date when the movie was released.
- `genres`: List of genres the movie belongs to.
- `production_companies`: List of production companies involved in the movie. ('-' separated production company)
- `credits`: List of the casts in the movie.
- `budget`: Budget allocated for the movie.
- `revenue`: Total revenue generated by the movie.
- `runtime`: Duration of the movie
- `vote_average`: Average vote or rating given by viewers
- `vote_count`: Total count of votes received for the movie.
- `popularity`: Popularity metric by TMDB.

````python
import pandas as pd
import numpy as np
# Display configuration
pd.options.display.max_colwidth = 40
pd.options.display.width = 0
pd.options.display.float_format = '{:,.2f}'.format

# 📊 Dataset Used in This Assignment (Size = 58 MB, gzip format)
data_url = "https://www.dropbox.com/scl/fi/eb2qypuy34gj8g051u71x/movies_dataset.gzip?rlkey=l3ua8qq0ffl7ezvfrhkovpzuo&st=nx8785hn&dl=0&raw=1"
df = pd.read_csv(data_url, compression='gzip')

print(df.head())
````

````python
# Part 1:
print("Movies with budget more than $100,000,000, sorted from highest budget to lowest budget")
#     Select only movies with positive vote_count and revenue > 10 million dollars)
df_filtered = df[(df['vote_count'] > 0) & (df['revenue'] > 100_000_000)]
df_sorted = df_filtered.sort_values(by='budget', ascending=False)
print(df_sorted[['title', 'budget']].head(20))

````

````python
import pandas as pd

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"

# Cleaning the data
df = pd.read_csv(data_url).dropna(subset=['total_sales']).reset_index(drop=True)

print(df)
````

````python
import pandas as pd

# Set the maximum width of each column to 25 characters, total 100 characters
pd.options.display.max_colwidth = 25
pd.options.display.width = 100

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url).dropna(subset=['total_sales'])

top_consoles = ['PC', 'PS2', 'DS', 'PS4', 'PS', 'NS',  'XOne', 'PS3', 'PSP', 'X360']
top_genres = ['Action', 'Adventure', 'Role-Playing', 'Shooter', 'Sports',
              'Strategy', 'Simulation', 'Racing', 'Puzzle', 'Platform', 'Fighting']

# Apply filtering
df = df[(df['console'].isin(top_consoles)) &
        (df['genre'].isin(top_genres))].reset_index(drop=True)

print(df)
````

## **2.6**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Combining DataFrames**
</div>

````python
import pandas as pd
import numpy as np

# Set the maximum width of each column to 25 characters, total 100 characters
pd.set_option('display.max_colwidth', 25)
pd.set_option('display.width', 100)

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url).replace(0, np.nan).sort_values(by='pal_sales')
df['publisher'] = df['publisher'].replace(r'Sony.*', 'Sony', regex=True)
df['genre'] = df['genre'].replace({'Action-Adventure': 'Act-Adv', 'Role-Playing':'RPG'})
df['title'] = df['title'] + ' (' + df['console'] + ')'

# Constructing dataframes for practice
df_ps3_games = df[df['console'] == 'PS3'].drop(columns=['console']).reset_index(drop=True).set_index('title')[['genre',	'publisher',	'developer']]
df_ps4_games = df[df['console'] == 'PS4'].drop(columns=['console']).reset_index(drop=True).set_index('title')[['genre',	'publisher',	'developer']]

df_ps3_ps4_games = df[df['console'].isin(['PS4', 'PS3'])].reset_index(drop=True).set_index('title')
df_ps3_ps4_games = df_ps3_ps4_games[~df_ps3_ps4_games.index.duplicated()]
df_ps3_ps4_games_west = df_ps3_ps4_games.dropna(subset=['na_sales'])[['na_sales','other_sales']]
df_ps3_ps4_games_asia = df_ps3_ps4_games.dropna(subset=['jp_sales'])[['jp_sales','pal_sales']]
df_ps3_ps4_games_critic = df_ps3_ps4_games.dropna()['critic_score']
````

````python
import pandas as pd
import numpy as np
# Display configuration
pd.options.display.max_colwidth = 40
pd.options.display.width = 0
pd.options.display.float_format = '{:,.2f}'.format

# 📊 Dataset Used in This Assignment (Size = 58 MB, gzip format)
data_url = "https://www.dropbox.com/scl/fi/eb2qypuy34gj8g051u71x/movies_dataset.gzip?rlkey=l3ua8qq0ffl7ezvfrhkovpzuo&st=nx8785hn&dl=0&raw=1"
df = pd.read_csv(data_url, compression='gzip')

print(df.head())
````

````python
# Part 1:
# Categorize movie's 'budget'. Movies are then sorted by the vote count
# Low Budget: <$1 million
# Moderate Budget: $1 million - $10 million
# Medium Budget: $10 million - $50 million
# High Budget: $50 million - $150 million
# Blockbuster Budget: >$150 million

# Create budget categories
budget_ranges = [0, 1000000, 10000000, 50000000, 150000000, float('inf')]
budget_labels = ['Low Budget', 'Moderate Budget', 'Medium Budget', 'High Budget', 'Blockbuster Budget']

# Categorize movies by budget
df['budget_cat'] = pd.cut(df['budget'], bins=budget_ranges, labels=budget_labels, right=False)

# Sort by vote_count in descending order
df_sorted = df.sort_values(by='vote_count', ascending=False)

print(df_sorted[['title', 'budget', 'budget_cat', 'vote_count']].head(20))
````

````python
# Part 2
# (Replace all 0 in the data with missing values first, then drop all rows with missing values)
# Number of Movies by Budget Category and Vote Average (convert vote average as integer first)
# Replace 0 values with NaN
df_clean = df.replace(0, np.nan)

# Drop rows with missing values
df_clean = df_clean.dropna()

# Convert vote_average to integer
df_clean['vote_average'] = df_clean['vote_average'].astype(int)

# Create a crosstab with vote_average as rows and budget_cat as columns
movie_counts = pd.crosstab(df_clean['vote_average'], df_clean['budget_cat'])

print(movie_counts)
````

````python
# Part 3
# Budget Categories as Dummy Variables. Then check the info (using .info())

# Create dummy variables from budget categories
budget_dummies = pd.get_dummies(df['budget_cat'])

# Concatenate the dummy variables with the title column from the original dataframe
df_with_dummies = pd.concat([df[['title']], budget_dummies], axis=1)

print(df_with_dummies.head(20))
print(df_with_dummies.info())
````

````python
# Part 4:
print("Log-scale QUANTILE Discretization of Movies Budget (sorted by Vote Count)")
# The categories name are the same as Part 1, but separate it by quantile
# As you cannot take log of 0, replace 0 with NaN first then drop NaN

# Replace 0 with NaN
df_log = df.copy()
df_log['budget'] = df_log['budget'].replace(0, np.nan)

# Apply log transformation to budget values
df_log['budget_log'] = np.log(df_log['budget'])

# Create quantile-based categories using qcut with same labels as before
df_log['budget_log_cat'] = pd.qcut(df_log['budget_log'], 
                                  q=5, 
                                  labels=budget_labels)


df_log_sorted = df_log.sort_values(by='vote_count', ascending=False)
print(df_log_sorted[['title', 'budget', 'budget_log', 'budget_log_cat']].head(20))
````

## **3.1**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Categorical Data**
</div>

````python
import pandas as pd
import numpy as np

# Set the maximum width of each column to 25 characters, total 100 characters
pd.set_option('display.max_colwidth', 25)
pd.set_option('display.width', 100)

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url).replace(0, np.nan).replace("Unknown",np.nan).\
        dropna(subset=['total_sales','critic_score']).reset_index(drop=True)

print(df)
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **3.1**.</font><font color="#0ea778" size='4'>**1**&nbsp;&nbsp;</font>


````python
# Categorize 'total_sales' into bins of 'Low', 'Medium', and 'High' sales
df['sales_category'] = pd.cut(df['total_sales'],
                                bins=[-np.inf, 0.5, 1.0, np.inf],
                                labels=['Low', 'Medium', 'High'])

print(df['sales_category'].head(), '\n')
print(df['sales_category'].value_counts())
````

````python
# Get all PlayStation games
ps_consoles = [console for console in df['console'].unique() if 'PS' in console]
df_ps = df[df['console'].isin(ps_consoles)].copy()

# Converting df_ps['console'] to a categorical data type with ordered categories.
df_ps['console'] = df_ps['console'].astype('category')

print(df_ps['console'].head(), '\n')
print(df_ps['console'].value_counts())
````

````python
# Define the desired order of categories
ps_console_order = ['PS', 'PS2', 'PS3', 'PSP', 'PSV', 'PS4', 'PSN']

# Reorder the categories (from 3.1.2)
df_ps['console'] = df_ps['console'].astype('category').cat.reorder_categories(ps_console_order)

print(df_ps['console'].cat.categories, '\n')
````

## **3.2**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Text Data**

````python
# Define the regex pattern for number of installment
pattern = r'\b(I{1,3}|II|III|IV|V|VI|VII|VIII|IX|XI{0,3}|XIV|XV|[1-9]|1[0-5])\b'

# Extract all the pattern from the 'title' then unstack(),
df_installment = df['title'].str.extractall(pattern)
df_installment = df_installment.unstack()

# Drop one level of the column index
df_installment.columns = df_installment.columns.droplevel(level=0)

# Drop the rows in df_installment so the result will have at least 2 non-NA values
df_2_installment = df_installment.dropna(thresh=2, axis=0)

# Perform inner join with the 'title' from df
df_title_installment = df[['title']].join(df_2_installment, how='inner')

# Generate new column names
new_columns_dict = {i:f'match_{i+1}' for i in range(len(df_title_installment.columns)-1)}

# Assign the new column names to the DataFrame
df_title_installment.columns = ['title'] + [new_columns_dict.get(i, col) for i, col in enumerate(df_title_installment.columns[1:])]

# Reset index and print
print(df_title_installment.reset_index(drop=True))
````

## **3.4**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Period Data**

````python
import pandas as pd
import numpy as np

# Set the maximum width of each column to 25 characters, total 100 characters
pd.set_option('display.max_colwidth', 40)
pd.set_option('display.width', 100)

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **3.4**.</font><font color="#0ea778" size='4'>**1**&nbsp;&nbsp;</font><br>

````python
# Convert 'release_date' into datetime object\n",
df['release_date'] = pd.to_datetime(df['release_date'])

# Make a new Period-column from df['release_date'] that has a daily period\n",
df['release_day'] = df['release_date'].dt.to_period('D')

print(df['release_day'])
````

````python
# Make a new Period-column from df['release_date']\n",
# that has a monthly period\n",
df['release_month'] = df['release_date'].dt.to_period('M')

# Get the number of games released in each console in monthly period\n",
df_console_monthly = df.groupby('release_month')['console']\
                       .value_counts().unstack().fillna(0).astype(int)

print(df_console_monthly.iloc[-5:,-5:])
````

````python
import pandas as pd
import numpy as np

# Set the maximum width of each column to 25 characters, total 100 characters
pd.set_option('display.max_colwidth', 40)
pd.set_option('display.width', 100)

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)

# Convert 'release_date' as datetime
df['release_date'] = pd.to_datetime(df['release_date'])
````

````python
# Make a timestamp for 2010-01-01
start_2010 = pd.Timestamp('2010-01-01')

# Make a new column with value 'Before 2010' or 'After 2010'
# depending on 'release date'
df['to_2010'] = df['release_date'].apply(lambda x: 'After 2010' if pd.notna(x) and x >= start_2010 else ('Before 2010' if pd.notna(x) and x < start_2010 else np.nan))

# Groupby 'to_2010' and calculate average sales for games released
sales_comparison_2010 = df.dropna(subset=['to_2010']).groupby('to_2010')['total_sales'].mean()

# To show the result
print(sales_comparison_2010)
````

````python
# Group by 'console' and find the first (min) and last (max) release dates
first_last_release = df.groupby('console')['release_date'].agg(['min', 'max'])

# Calculate the time difference between the first and last release
first_last_release['time_span'] = first_last_release['max'] - first_last_release['min']

# Drop rows with missing values and sort by 'time_span' (descending)
first_last_release = first_last_release.dropna().sort_values(by='time_span', ascending=False)

print(first_last_release[['min', 'max', 'time_span']].head(10))
````

````python
# Get all rows with publisher "Nintendo", sort by release_date, and reset the index
df_nin = df[df['publisher']=='Nintendo'].sort_values('release_date', ignore_index=True).copy()

# Calculate the time difference between release date
df_nin['release_date_diff'] = df_nin['release_date'].diff()

# Calculate the age of each game from its release date until now
today_timestamp = pd.Timestamp.now()
df_nin['age'] = today_timestamp - df_nin['release_date']

# Drop all missing values
df_nin = df_nin.dropna(subset=['release_date', 'age'])

# Decompose timedelta into years, months, and days (use floor division '//')
df_nin['years'] = df_nin['age'].dt.days // 365
df_nin['months'] = (df_nin['age'].dt.days % 365) // 30
df_nin['days'] = (df_nin['age'].dt.days % 365) % 30

# Create the 'age_long' column
df_nin['age_long'] = "Years: " + df_nin['years'].astype(str).str.rjust(2) + ", Month: " + df_nin['months'].astype(str).str.rjust(2) + ", Day: " + df_nin['days'].astype(str).str.rjust(2)

print(df_nin[['title', 'age', 'age_long']].sort_values('age'))
````

## **4.1**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Basic Plot Types**

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#%config InlineBackend.figure_format = 'retina'  # For higher resolution
plt.rcParams['figure.figsize'] = [6, 4]  # Default figure size

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)

# Convert 'release_date' as datetime
df['release_date'] = pd.to_datetime(df['release_date'])
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.1**.</font><font color="#0ea778" size='4'>**1**&nbsp;&nbsp;</font>

````python
# Get all rows with the console is in PS console list
df_ps = df[df['console'].isin(['PS','PSP','PSV','PS2','PS3','PS4'])]

# Group by console, calculate the sum of 'total_sales'
df_sales_ps = df_ps.groupby('console')['total_sales'].sum()
#print(df_sales_ps)

# Making vertical bar graph
plt.bar(df_sales_ps.index, df_sales_ps.values)

plt.title('Sum of Total Sales for PS Consoles')
plt.xlabel('Console')
plt.ylabel('Total Sales')
plt.show()
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.1**.</font><font color="#0ea778" size='4'>**2**&nbsp;&nbsp;</font><br>



````python
# Get all rows with the publisher is "Nintendo"
df_nin = df[df['publisher'] == 'Nintendo'].copy()
df_nin['release_date'] = pd.to_datetime(df_nin['release_date'], errors='coerce')

# Resample the 'release_date' by year and the sum of 'total_sales'
yearly_sales = df_nin.resample('Y', on='release_date')['total_sales'].sum()

# Set figure size to be 8 by 3
plt.figure(figsize=(8, 3))

# Make line graph for yearly total sales
plt.plot(yearly_sales.index.year, yearly_sales.values)

plt.title('Yearly Total Sales of Nintendo Games')
plt.xlabel('')
plt.ylabel('Total Sales')
plt.show()
````

## **4.2**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Colors and Style Customization**

````python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
mpl.rcParams.update(mpl.rcParamsDefault)
#%config InlineBackend.figure_format = 'retina'  # For higher resolution
plt.rcParams['figure.figsize'] = [6, 4]  # Default figure size

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)

# Convert 'release_date' as datetime
df['release_date'] = pd.to_datetime(df['release_date'])
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.2**.</font><font color="#0ea778" size='4'>**1**&nbsp;&nbsp;</font><br>

## **4.4**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Annotations & Legends**

````python
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
mpl.rcParams.update(mpl.rcParamsDefault)
#%config InlineBackend.figure_format = 'retina'  # For higher resolution
plt.rcParams['figure.figsize'] = [6, 4]  # Default figure size

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)

# Convert 'release_date' as datetime
df['release_date'] = pd.to_datetime(df['release_date'])
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.4**.</font><font color="#0ea778" size='4'>**1**&nbsp;&nbsp;</font><br>

````python
# Get all rows with the genre is in the given list
df_genre = df[df['genre'].isin(['Role-Playing', 'Adventure', 'Shooter'])].copy()

# Make a column for release year
df_genre['release_year'] = df_genre['release_date'].dt.year

# Aggregate number of titles released by year and genre
release_trends = df_genre.groupby(['release_year', 'genre'])['title'].count().unstack()

# Plot each genre individually
plt.figure(figsize=(9, 5))

# Create subplots
plt.plot(release_trends.index, release_trends['Role-Playing'],  marker='.', label='Role-Playing')
plt.plot(release_trends.index, release_trends['Adventure'], marker='.', label='Adventure')
plt.plot(release_trends.index, release_trends['Shooter'],  marker='.',label='Shooter')

# Customize the plot
plt.title('Popular Genre Release (1990-2020)')
plt.xlabel('')
plt.xlim(1990,2021)
plt.xticks(np.arange(1990, 2021,2))
plt.ylabel('Number of Released Game')
plt.grid(True)

# Add a legend
plt.legend(title='Genre')

# Add text 'Transition to New Consoles' at (2011, 20)
plt.text(2011, 20, 'Transition to New Consoles')
plt.show()
````

````python

# Map the related company into "Bandai", "Namco", and "Bandai Namco"\
bandai_namco_mapping = {\
    'Bandai': 'Bandai',\
    'Bandai Namco': 'Bandai Namco',\
    'Bandai Namco Entertainment': 'Bandai Namco',\
    'Bandai Namco Games': 'Bandai Namco',\
    'Bandai Visual': 'Bandai',\
    'Namco': 'Namco',\
    'Namco Bandai': 'Bandai Namco',\
    'Namco Bandai Games': 'Bandai Namco',\
    'Namco Networks America': 'Namco',\
    'Namco Networks America Inc.': 'Namco',\
    'Namco Tales Studio / Wolfteam': 'Namco',\
    'Wolfteam / Namco Tales Studio': 'Namco'\
}\

# Filter relevant publishers and make a copy
df_bn = df[df['publisher'].str.contains('Bandai|Namco|Bandai Namco', na=False)].copy()

# Replace the publisher names based on the mapping
df_bn['publisher'] = df_bn['publisher'].replace(bandai_namco_mapping)

# Make a column for release year
df_bn['release_year'] = df_bn['release_date'].dt.year

# Aggregate sales data by year and publisher
release_trends = df_bn.groupby(['release_year', 'publisher'])['title'].count().unstack().replace(0,np.nan)

# Set figure size
plt.figure(figsize=(9, 5))

# Plot individually (all default styles other than line style, marker type, and label)
plt.plot(release_trends.index, release_trends['Bandai'],  marker='.', label='Bandai')
plt.plot(release_trends.index, release_trends['Namco'], marker='.', label='Namco')
plt.plot(release_trends.index, release_trends['Bandai Namco'],  marker='.',label='Bandai Namco')

plt.title('Bandai Namco Released Games')
plt.xlabel('Year')
plt.ylabel('Number of Released Game')
plt.legend(title='Publisher')

# Add a vertical line at x=2006 with dashed-style
plt.axvline(x=2006, color='gray', linestyle='--')

# Add text annotation at (2006, 100)
plt.text(2006, 100, 'Acquisition')

plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
````

````python
# Get all rows with the publisher is "Zynga"
df_zynga = df[df['publisher'].str.contains('Zynga')].copy()

# Resample the 'release_date' by year and count the number of 'title'
yearly_release = df_zynga.resample('Y', on='release_date')['title'].count()

# Ensure the x-axis includes 2008 even if the data is 0
yearly_release = yearly_release.reindex(pd.date_range('2008-01-01', '2017-12-31', freq='Y'), fill_value=0)

# Make bar graph for yearly released game
plt.figure(figsize=(8, 4))
bars = plt.bar(yearly_release.index.year, yearly_release, color='indianred')

# Customize the plot
plt.title('Number of Games Released by Zynga')
plt.xlim(2009, 2017)

# Add the value inside the bars with this customization:
# ha='center', va='center', color='white', fontsize='12', fontweight='bold'
for bar in bars:
    if bar.get_height() > 0:  # Only add text for non-zero bars
        plt.text(bar.get_x() + bar.get_width() / 2., bar.get_height() / 2,
                 int(bar.get_height()), ha='center', va='center', color='white',
                 fontsize='12', fontweight='bold')

# Set the ticks and labels
plt.xticks(yearly_release.index.year)
plt.yticks([])
plt.tick_params(axis='y', left=False, labelleft=False)
plt.xlabel('Year')
plt.ylabel('')

# Hide all the spines
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['bottom'].set_visible(False)
plt.gca().spines['left'].set_visible(False)

plt.show()
````

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib as mpl
import matplotlib.pyplot as plt
mpl.rcParams.update(mpl.rcParamsDefault)
#%config InlineBackend.figure_format = 'retina'  # For higher resolution
plt.rcParams['figure.figsize'] = [6, 4]  # Default figure size

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)

# Convert 'release_date' as datetime
df['release_date'] = pd.to_datetime(df['release_date'])
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.5**.</font><font color="#0ea778" size='4'>**1**&nbsp;&nbsp;</font><br>

````python
# Get all rows with the genre is in the given list
df_genre = df[df['genre'].isin(['Role-Playing', 'Adventure', 'Shooter'])].copy()

# Make a column for release year
df_genre['release_year'] = df_genre['release_date'].dt.year

# Aggregate number of titles released by year and genre
release_trends = df_genre.groupby(['release_year', 'genre'])['title'].count().unstack()

# Plot each genre individually
plt.figure(figsize=(7, 7))

# Create subplots
plt.subplot(3, 1, 1)
plt.plot(release_trends.index, release_trends['Role-Playing'],
         marker='.', color='slateblue')
plt.title('Number of Released\\nRole-Playing Games')

plt.subplot(3, 1, 2)
plt.plot(release_trends.index, release_trends['Adventure'],
         marker='.', color='orange')
plt.title('Adventure Games')

plt.subplot(3, 1, 3)
plt.plot(release_trends.index, release_trends['Shooter'],
         marker='.', color='green')
plt.title('Shooter Games')

plt.tight_layout()
plt.show()
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.5**.</font><font color="#0ea778" size='4'>**2**&nbsp;&nbsp;</font><br>

````python
# Get all rows with the genre is in the given list
df_genre = df[df['genre'].isin(['Role-Playing', 'Adventure', 'Shooter'])].copy()

# Make a column for release year
df_genre['release_year'] = df_genre['release_date'].dt.year

# Aggregate number of titles released by year and genre
release_trends = df_genre.groupby(['release_year', 'genre'])['title'].count().unstack()

# Plot each genre individually
plt.figure(figsize=(6, 6))

# Create first subplot
ax1 = plt.subplot(3, 1, 1)
plt.plot(release_trends.index, release_trends['Role-Playing'],
         marker='.', color='slateblue')
plt.title('Number of Released\nRole-Playing Games')

# Set x-axis limits for the first plot (1995-2020)
ax1.set_xlim(1995, 2020)

# Create second subplot, sharing the same axis as ax1
ax2 = plt.subplot(3, 1, 2, sharex=ax1)
plt.plot(release_trends.index, release_trends['Adventure'],
         marker='.', color='orange')
plt.title('Adventure Games')

# Create third subplot, sharing the same axis as ax2
ax3 = plt.subplot(3, 1, 3, sharex=ax2)
plt.plot(release_trends.index, release_trends['Shooter'],
         marker='.', color='green')
plt.title('Shooter Games')

plt.tight_layout()
plt.show()
````

[➤]

<img src="https://i.imgur.com/Lnxu6v7.png" width=550 />

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.5**.</font><font color="#0ea778" size='4'>**3**&nbsp;&nbsp;</font><br>

````python
# Get all rows with publisher is Nintendo with release_date between 2010 and 2020
# with selected genres
condition_1 = df['publisher'] == 'Nintendo'
condition_2 = df['release_date'] >= pd.Timestamp('2010-01-01')
condition_3 = df['release_date'] <= pd.Timestamp('2020-12-31')
condition_4 = df['jp_sales'] > 0.01
df_nin = df[condition_1 & condition_2 & condition_3 & condition_4].sort_values('jp_sales')

# Plot the distribution of jp_sales by genre
plt.figure(figsize=(6, 6))
sns.boxplot(x='jp_sales', y='genre', data=df_nin)

plt.title('Distribution of Japanese Sales by Genre for Nintendo (2010-2020)') # Corrected year range
plt.xlabel('Sales in Japan (millions)')
plt.ylabel('Genre')
plt.show()
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **4.5**.</font><font color="#0ea778" size='4'>**4**&nbsp;&nbsp;</font><br>


````python
# Get all rows with publisher is Nintendo with release_date between 2010 and 2020
# with selected genres
condition_1 = df['publisher'] == 'Nintendo'
condition_2 = df['release_date'] >= pd.Timestamp('2010-01-01')
condition_3 = df['release_date'] <= pd.Timestamp('2020-12-31')
condition_4 = df['genre'].isin(['Adventure', 'Action', 'Role-Playing'])
df_nin = df[condition_1 & condition_2 & condition_3 & condition_4].copy()

# Define color palette for each category
genre_colors = {'Adventure': 'hotpink', 'Action': 'navy', 'Role-Playing': 'limegreen'}

# Plot the pairplot
# columns: 'na_sales','jp_sales','pal_sales', 'other_sales'
# color: 'genre'
sns.pairplot(df_nin[['na_sales','jp_sales','pal_sales', 'other_sales', 'genre']],
             hue='genre', palette=genre_colors, corner=True, height=2, aspect=1)

plt.show()
````

````python
# 4.5.5
# Get all rows with publisher is Nintendo with release_date between 2000 and 2020
# with selected genres
condition_1 = df['publisher'] == 'Nintendo'
condition_2 = df['release_date'] >= pd.Timestamp('2000-01-01')
condition_3 = df['release_date'] <= pd.Timestamp('2020-12-31')
condition_4 = df['genre'].isin(['Adventure', 'Action', 'Role-Playing'])
condition_5 = df['jp_sales'] > 0.01
df_nin = df[condition_1 & condition_2 & condition_3 & condition_4 & condition_5].copy()

# Categorizing 'jp_sales' into 'low', 'mid', 'high' based on quantiles
df_nin['jp_sales_cat'] = pd.qcut(df_nin['jp_sales'],
                                 q=3, labels=['low', 'mid', 'high'], duplicates='drop') # Added duplicates='drop'
df_nin['release_date_year'] = df_nin['release_date'].dt.year

# Use facet grid to make scatter plot with
# columns: 'genre', row: 'jp_sales_cat'
# x-axis: 'release_date_year', y-axis: 'total_sales'
g = sns.FacetGrid(df_nin, col='genre', row='jp_sales_cat', height=2, aspect=1.5)
g.map(plt.scatter, 'release_date_year', 'total_sales')  # Specify both x and y variables for the scatter plot
g.set_xticklabels(rotation=90)


plt.tight_layout()
plt.show()
````

````python
# 4.5.6
# Get all rows with the genre is in the given list
df_genre = df[df['genre'].isin(['Role-Playing', 'Adventure', 'Shooter', 'Action'])].copy()

# Make a column for release year
df_genre['release_year'] = df_genre['release_date'].dt.year

# Aggregate number of titles released by year and genre
release_trends = df_genre.groupby(['release_year', 'genre'])['title'].count().unstack()

# Plot each genre individually
plt.figure(figsize=(9, 6))

# Create subplots
plt.subplot(2, 2, 1) # 2 rows, 2 column, 1st subplot
plt.plot(release_trends.index, release_trends['Role-Playing'],
         marker='.', color='slateblue')
plt.title('Role-Playing Games')

plt.subplot(2, 2, 2) # 2 rows, 2 column, 2nd subplot
plt.plot(release_trends.index, release_trends['Adventure'],
         marker='.', color='orange')
plt.title('Adventure Games')

plt.subplot(2, 2, 3) # 2 rows, 2 column, 3rd subplot
plt.plot(release_trends.index, release_trends['Shooter'],
         marker='.', color='green')
plt.title('Shooter Games')

plt.subplot(2, 2, 4) # 2 rows, 2 column, 4th subplot
plt.plot(release_trends.index, release_trends['Action'], # Corrected to Action
         marker='.', color='red')
plt.title('Action Games')

plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display configuration
pd.options.display.max_colwidth = 70
pd.options.display.width = 0
pd.options.display.float_format = '{:,.2f}'.format
plt.rcParams['figure.figsize'] = (9, 5)

# 📊 Dataset Used in This Assignment (Size = 58 MB, gzip format)
data_url = "https://www.dropbox.com/scl/fi/eb2qypuy34gj8g051u71x/movies_dataset.gzip?rlkey=l3ua8qq0ffl7ezvfrhkovpzuo&st=nx8785hn&dl=0&raw=1"
df = pd.read_csv(data_url, compression='gzip').replace(0, np.nan).dropna()

print(df.head())
````

````python
# Filter for movies with 'Disney' in production_companies
df_disney = df[df['production_companies'].str.contains('Disney', na=False)].copy()

# Convert release_date to datetime and extract year
df_disney['release_date'] = pd.to_datetime(df_disney['release_date'], errors='coerce')
yearly_releases = df_disney.groupby(df_disney['release_date'].dt.year)['title'].count().dropna()

# Plot a bar chart
plt.figure(figsize=(10, 5))
plt.bar(yearly_releases.index, yearly_releases.values)

# Customize plot
plt.title('Number of Movies Released by Disney Each Year', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Number of Movies')
plt.grid(axis='y', alpha=0.5)
plt.xticks(np.arange(int(yearly_releases.index.min()), int(yearly_releases.index.max())+1, 5)) # Set x-ticks every 5 years

plt.show()
````

[➤]

<img src="https://i.imgur.com/PLeYdoj.png" width=650 />

````python
# Filter movies containing 'Pixar' or 'DreamWorks'
df_pixar = df[df['production_companies'].str.contains('Pixar', na=False)].copy()
df_dream = df[df['production_companies'].str.contains('DreamWorks', na=False)].copy()

# Convert release_date to datetime
df_pixar['release_date'] = pd.to_datetime(df_pixar['release_date'], errors='coerce')
df_dream['release_date'] = pd.to_datetime(df_dream['release_date'], errors='coerce')

# Resample by year and sum revenue, then convert to billions
yearly_revenue_pixar = df_pixar.resample('Y', on='release_date')['revenue'].sum() / 1e9
yearly_revenue_dream = df_dream.resample('Y', on='release_date')['revenue'].sum() / 1e9

# Plot line chart
plt.figure(figsize=(10, 6))

# Add bar charts with alpha=0.1
plt.bar(yearly_revenue_pixar.index.year, yearly_revenue_pixar.values, alpha=0.1)
plt.bar(yearly_revenue_dream.index.year, yearly_revenue_dream.values, alpha=0.1)

# Add line charts
plt.plot(yearly_revenue_pixar.index.year, yearly_revenue_pixar.values, marker='o', label='Pixar (Line)')
plt.plot(yearly_revenue_dream.index.year, yearly_revenue_dream.values, marker='s', label='DreamWorks (Line)')

# Customize plot
plt.title("Total Yearly Revenue of Pixar and DreamWorks Movies", fontsize=14)  # Match output title
plt.xlabel("Year")
plt.ylabel("Revenue (Billions USD)")  # Match output label
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
````

[➤]

image.png

````python

# Filter for movies containing 'Harry Potter'
df_hp = df[df['title'].str.contains('Harry Potter', na=False)].copy()

# Convert release_date to datetime and sort
df_hp['release_date'] = pd.to_datetime(df_hp['release_date'], errors='coerce')
df_hp_sorted = df_hp.sort_values('release_date').set_index('title') # Set index for plotting

# Plot budget and revenue as horizontal bar charts
ax = df_hp_sorted['revenue'].plot(kind='barh', figsize=(12, 6), color='midnightblue', label='Revenue')
df_hp_sorted['budget'].plot(kind='barh', color='gold', label='Budget', ax=ax)

# Customize plot
plt.title('"Harry Potter" Movies Budget & Revenue (Oldest to Newest)', fontsize=14)
plt.ylabel("Movie Title")
plt.xlabel("Amount (USD)")
plt.legend(title='Metric')
plt.grid(axis='x', alpha=0.5)

plt.tight_layout() # Adjust layout to prevent labels overlapping
plt.show()
````

````python
df_list_title = df.copy()

# Ensure 'release_date' is a datetime type
df_list_title['release_date'] = pd.to_datetime(df_list_title['release_date'])

# Make a new column 'title-console'
df_list_title['title_console'] = df_list_title['title'] + ' (' + df_list_title['console'] + ')'

# Resample the DataFrame to fill in all dates, aggregating titles into lists
resampled_df = df_list_title.resample('D', on='release_date')['title_console']\
                 .agg(lambda x: list(x) if not x.empty else []).reset_index()

# Save it as excel file (Sheet name is 'Sheet1')
resampled_df.to_excel('release_date_title.xlsx', sheet_name='Sheet1', index=False)


# Check if it is made correctly
!ls -lh
````

````python
import numpy as np
import pandas as pd
pd.reset_option('^display.')

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)
````

````python
# Optimizing (Select only necessary columns & set data types)
df_ff = df[['title', 'console', 'total_sales', 'release_date']].copy()
df_ff['console'] = df_ff['console'].astype('category')
df_ff['total_sales'] = df_ff['total_sales'].astype('float32')

# This is the answer of 3.4.10 but the columns are optimized first
# Check the time (CPU and Wall time) used for each steps (every line that starts with df_ff)
print("Filter to include only 'Final Fantasy' titles")
%time
df_ff = df_ff[df_ff['title'].str.contains('Final Fantasy', na=False)].copy()

print("\nConvert 'release_date' to datetime")
%time
df_ff['release_date'] = pd.to_datetime(df_ff['release_date'])

print("\nMake a yearly period 'release_year' from 'release_date'")
%time
df_ff['release_year'] = df_ff['release_date'].dt.to_period('Y')

print("\nSort by 'release_year' (ascending) and 'total_sales' (descending)")
%time
df_ff = df_ff.sort_values(['release_year', 'total_sales'], ascending=[True, False])

print("\nDrop the duplicates 'release_year'")
%time
df_ff = df_ff.drop_duplicates(subset='release_year')

print("\nSet 'release_year' as the index (drop the original index)")
%time
df_ff = df_ff.set_index('release_year', drop=True)

print("\nPrint the titles")
%time
print(df_ff[['title', 'console']].head())
````

## **5.6**<font color="#F1A006">   ▝</font><font color="#707DA0">▖   </font>**Parallel Processing with Pandas**

````python
import os
import numpy as np
import pandas as pd
pd.reset_option('^display.')

# 📊 Dataset Used in This Exercise
data_url = "https://www.dropbox.com/scl/fi/g6infpafqxcb76i5w15mr/vg-chartz-2024.csv?rlkey=ecdupx82jexit05ma6n6f2nru&st=zud8fpyx&dl=0&raw=1"
df = pd.read_csv(data_url)

# Ensure 'release_date' is a datetime type
df['release_date'] = pd.to_datetime(df['release_date'])
````

<font color="#0ea778" size='4'>❯</font><font color="#9A9A9A" size='2'>❯</font><font color="#9A9A9A" size='2'>❯</font><font size='2'> **5.6**.</font><font color="#0ea778" size='4'>**1**&nbsp;&nbsp;</font>

````python
from joblib import Parallel, delayed
import pandas as pd

# Write a function that calculates the mean total sales for each genre
def calculate_average_sales(df_input, genre):
    genre_data = df_input[df_input['genre'] == genre]
    return genre_data['total_sales'].mean()

# Reading file
df = pd.read_csv(data_url)

# Do parallel computing with 4 jobs to calculate mean for each genre
genres = df['genre'].unique()
results = Parallel(n_jobs=4)(delayed(calculate_average_sales)(df, genre) for genre in genres)

# Print the result
print(pd.Series(dict(zip(genres, results))))
````

````python
from joblib import Parallel, delayed

# Write a function that calculates the median critic score for each console
def aggregate_reviews(data, console):
    platform_data = data[data['console'] == console]
    return platform_data['critic_score'].median()

# Reading file
df = pd.read_csv(data_url)

# Do parallel computing with 8 jobs (verbose=10) to calculate mean for each genre
platforms = df['console'].unique()
results = Parallel(n_jobs=8, verbose=10)(delayed(aggregate_reviews)(df, platform) for platform in platforms)

# Print the result
print(pd.Series(dict(zip(platforms, results))))
````

````python
from joblib import Parallel, delayed

# Write a function that calculates the mean sales for each genre in a given console
def groupby_console(data, console):
    df_console = data[data['console'] == console].drop(columns='console')
    df_console_groupby = df_console.groupby('genre').agg({'total_sales': 'mean', 'na_sales': 'mean', 'jp_sales': 'mean', 'pal_sales': 'mean'})
    return df_console_groupby

# Reading file
df = pd.read_csv(data_url)

# Do parallel computing with 8 jobs to calculate mean for each genre
consoles = df['console'].unique()
results = Parallel(n_jobs=8)(delayed(groupby_console)(df, console) for console in consoles)


# Print the result using a DataFrame for better readability
result_dict = dict(zip(consoles, results))
result_df = pd.concat(result_dict, axis=0)
print(result_df)
````

````python
# Part 2
# Read movie_dataset_1.csv, set dtypes accordingly, and check the memory (ignore warning if any)
# Define dtypes to optimize memory usage
# Based on potential data ranges and types
dtype_mapping = {
    'budget': 'float16',        # Financial values, likely many zeros, float16 is sufficient
    'revenue': 'float16',       # Financial values, likely many zeros, float16 is sufficient
    'runtime': 'float32',       # Runtime values, float32 is usually enough precision
    'vote_average': 'float32',  # Rating scores, float32 is usually enough precision
    'vote_count': 'float16',    # Count values, float16 if max count is within range, might have NaNs
    'popularity': 'float64'     # Popularity scores can be high and require more precision
}

# Columns to parse as dates
parse_dates = ['release_date']

# Read the first chunk file with specified dtypes and date parsing

df_1 = pd.read_csv("movie_dataset_1.csv",
                    dtype=dtype_mapping,
                    parse_dates=parse_dates)

# Convert 'genres' column to category type after reading
# Category is efficient if the number of unique genre combinations is much less than total rows
df_1['genres'] = df_1['genres'].astype('category')

# Get memory usage for each column
mem_usage = df_1.memory_usage(deep=True)
dtypes = df_1.dtypes

# Create a summary dataframe for printing
memory_df = pd.DataFrame({
    'DType': dtypes,
    'Memory_Usage': mem_usage
})

# Format memory usage for display (convert bytes to KB)
memory_df['Memory_Usage'] = memory_df['Memory_Usage'] / 1024 # Convert to KB
memory_df['Memory_Usage'] = memory_df['Memory_Usage'].map('{:.2f} KB'.format) # Format to 2 decimal places KB

# Print the memory usage dataframe, ensuring the index name is as in the example
memory_df.index.name = 'Index'
print(memory_df)
````
